```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
def load_data(file_path):
    tokens = []
    labels = []
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line:
                parts = line.split()
                if len(parts) >= 2:
                    tokens.append(parts[0])
                    labels.append(parts[1])
    return tokens, labels
```

```python
train_tokens, train_labels = load_data('train_corrected.txt')
test_tokens, test_labels = load_data('test_corrected.txt')

vectorizer = TfidfVectorizer(lowercase=False, token_pattern=r"\S+")
X_train = vectorizer.fit_transform(train_tokens)
X_test = vectorizer.transform(test_tokens)
```

```python
print("Training SVM...")
svm = LinearSVC(dual='auto', max_iter=200)
svm.fit(X_train, train_labels)

print("\nPredicting...")
predictions = svm.predict(X_test)
unique_labels = sorted(list(set(test_labels)))
```

```python
print(f"Accuracy Score: {accuracy_score(test_labels, predictions) * 100:.2f}%")
print(classification_report(test_labels, predictions, digits=4, zero_division=0))
```
